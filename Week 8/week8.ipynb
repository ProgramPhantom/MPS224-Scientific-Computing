```python
def gasses_only(df):
    """
    Return a DataFrame containing only the rows where the 'Phase' is 'Gas'.
    """
    # Remake dataframe so index is 0, 1, 2, ...
    return df[df["Phase"] == "gas"][["Element", "Symbol"]].reset_index(drop=True)

gasses = gasses_only(data)
gasses
```

```python
def highest_bp_element(df):
    """`
    return the name of the element that has the highest boiling point
    """
    return df.loc[df["Boiling Point (K)"].idxmax()]["Element"]

el = highest_bp_element(data)
el
```

```python
def Ds_stripped(df):
    """
    pandas series containing information about the element Darmstadtium.
    """
    ds_row = df[df["Element"] == "Darmstadtium"].iloc[0]
    return ds_row.dropna()


ds_info = Ds_stripped(data)
ds_info
```

```python
omega = np.array([0.1 * i for i in range(9)])

def p(t, a0, a1, a2, a3, a4, a5, a6, a7, a8, a9, a10, a11, a12, a13, a14, a15, a16, a17, a18):
    """
    Compute p for each time t using the given coefficients a0, a1, ..., a18 and the global variable omega.
    """
    global omega

    def p_t(t):
        cos_terms = np.array([a1, a3, a5, a7, a9, a11, a13, a15, a17]) * np.cos(omega * t)
        sin_terms = np.array([a2, a4, a6, a8, a10, a12, a14, a16, a18]) * np.sin(omega * t)
        return a0 + np.sum(cos_terms) + np.sum(sin_terms)

    res = np.vectorize(p_t)(t)
    
    if res.size == 1:
        return float(res)
    return res    

result = p([0, 1], *[0]*19)
result
```
